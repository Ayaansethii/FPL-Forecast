```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error

model_df = pd.read_parquet("../data/processed/features.parquet")
TEST_SEASON = "2025-26"
test = model_df[model_df["season"] == TEST_SEASON]
print(test.shape)
```

```python
actual = test["total_points"]

zero_pred = np.zeros(len(test))
mae_zero = np.mean(np.abs(zero_pred - actual))

mae_last5 = mean_absolute_error(actual, test["total_points_avg_last_5"])

mae_xp = mean_absolute_error(actual, test["xP"])

print(f"Predict 0:      {mae_zero:.3f}")
print(f"Last-5 average: {mae_last5:.3f}")
print(f"FPL xP:         {mae_xp:.3f}")
```

## Baseline results (test season: 2025-26)

| Baseline | MAE |
|---|---|
| Predict 0 for everyone | 1.162 |
| Last-5 match average | **1.042** |
| FPL expected points (xP) | 1.061 |

**Target to beat:** MAE below 1.042 (last-5 average), the best baseline. Interestingly, it slightly outperformed FPL's own xP.
